```python
import matplotlib
import utils
import os
import numpy as np
import nibabel as nib
```

```python
# load input data
input_dir = 'data/'

# list all .mgz files in the input directory
mgz_files = [f for f in os.listdir(input_dir) if f.endswith('.mgz')]

# load files
brains = []
for mgz_file in mgz_files:
    data = nib.load(input_dir + mgz_file)
    data = data.get_fdata()
    print(f'Loaded {mgz_file} with shape {data.shape}')
    brains.append(data)

print(f'Total brains loaded: {len(brains)}')
```

```python
print("shape of first brain:", brains[0].shape)
```

```python
# visualize the first brain
utils.plot3Views(brains[0], vmin=0, vmax=199, slice1=128, cmap='viridis', title='3D Views')
```

```python
# load the outputs 
output_dir = 'outFiles/'
npy_files = [f for f in os.listdir(output_dir) if f.endswith('.npy')]
predictions = []
for npy_file in npy_files:
    pred = np.load(output_dir + npy_file)
    print(f'Loaded {npy_file} with shape {pred.shape}')
    predictions.append(pred)    
    
```

```python
# visualize the first brain
utils.plot3Views(predictions[0], vmin=0, vmax=199, slice1=64, cmap='viridis', title='Predicted 3D Views')
```
